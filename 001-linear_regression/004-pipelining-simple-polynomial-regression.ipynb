```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
# random generator
rng = np.random.default_rng(seed=42)
```

```python
X = 6 * rng.random((100,1)) - 3
y = 2 + (1.5*X) + (0.5 * (X**2)) + rng.random((100,1))
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=42,test_size=0.3)
```

```python
def poly_regression(degree=2):
    X_test = np.linspace(-3,3,200).reshape(-1,1)

    poly_feature = PolynomialFeatures(degree=degree)
    lin_reg = LinearRegression()

    poly_reg = Pipeline([
        ("poly_features" ,poly_feature),
        ("lin_reg",lin_reg)
    ])

    # here you dont have to transform into polynomial features by yourself, pipeline does it for us and also add that polynomial features
    # into linear regressiojn model
    poly_reg.fit(X_train,y_train)
    # X_test as well we dont have to transform manually
    y_pred = poly_reg.predict(X_test)

   
    sns.scatterplot(x=X_train[:,0], y=y_train[:,0], label="Training Data")
    sns.lineplot(x=X_test.ravel(), y=y_pred.ravel(), color="red", label="Polynomial Curve")
   
    
```

```python
poly_regression(degree=1)
```

```python
poly_regression(degree=2)
```
