### 1. Imports

```python
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.datasets import fetch_california_housing

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import (
    LinearRegression,
    Ridge,
    Lasso,
    ElasticNet
)

from sklearn.metrics import (r2_score, mean_squared_error)
```

### 2. Loading Dataset

```python
housing = fetch_california_housing()
```

```python
housing.keys()
```

```python
df = pd.DataFrame(data=housing['data'],columns=housing['feature_names'])
df['Price'] = housing['target']
```

```python
df.head()
```

### 3. Understanding the data

```python
df.shape
```

```python
df.info()
```

```python
# any missing values
df.isnull().sum()
```

```python
df.describe().T
```

### 4. Seprate Feature and Target

```python
X = df.drop(columns="Price")
y = df["Price"]
```

```python
X.shape , y.shape
```

### 5. Train-Test Split

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=42,test_size=0.2)
```

### 6. Standardization (so that all the features comes to the same scale)

```python
print(X_train.shape ,X_test.shape) 
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
y_train.to_numpy().reshape(-1,1)
y_test.to_numpy().reshape(-1,1)
```

### 7. Training the baseline model

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
print("MSE :", mean_squared_error(y_test, y_pred))
print("R²  :", r2_score(y_test, y_pred))
```

```python
coef_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": lr.coef_
})

coef_df
```

## 1. Ridge Regression

```python
ridge = Ridge(alpha=20.0)
ridge.fit(X_train,y_train)
ridge_pred = ridge.predict(X_test)
```

```python
print("MSE :", mean_squared_error(y_test, ridge_pred))
print("R² :", r2_score(y_test, ridge_pred))
```

```python
#### Comparing coefficeint 
```

```python

coef = pd.DataFrame({
    "Feature" : X.columns,
     "Liner Coef" : lr.coef_,
    "Ridge Coef" : ridge.coef_
})
```

```python
coef
```

## 2. Lasso

```python
# let's try with lasso
lasso = Lasso(alpha=0.1)
lasso.fit(X_train,y_train)
lasso_pred = lasso.predict(X_test)
```

```python
print("MSE :", mean_squared_error(y_test, lasso_pred))
print("R² :", r2_score(y_test, lasso_pred))
```

```python

coef = pd.DataFrame({
    "Feature" : X.columns,
     "Liner Coef" : lr.coef_,
    "Ridge Coef" : ridge.coef_,
    "Lasso Coef" : lasso.coef_
})
coef
```

Lasso performs feature selection by shrinking coefficients of less important features all the way to exactly zero.

It's not necessarily that features are completely useless before applying Lasso; rather, when features are correlated or weak predictors, Lasso zeroes out redundant ones to simplify the model.

## 3. Elastic Net

```python
elastic_net = ElasticNet(alpha=0.03 , l1_ratio=0.2)
elastic_net.fit(X_train,y_train)
elastic_pred = elastic_net.predict(X_test)
```

```python
print("MSE :", mean_squared_error(y_test, elastic_pred))
print("R² :", r2_score(y_test, elastic_pred))
```

```python
coef = pd.DataFrame({
    "Feature" : X.columns,
     "Liner Coef" : lr.coef_,
    "Ridge Coef" : ridge.coef_,
    "Lasso Coef" : lasso.coef_,
    "Elastic coef" : elastic_net.coef_
})
coef
```
