## Data Cleaning

```python
# missing values
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
# We are storing Sidi-Bel Abbes Region Dataset as 1 as region  as below 122 all belong to it 
```

```python
df.loc[:122,"Region"] = 0
df.loc[122:,"Region"] = 1
# conver the dtype of Region to int as it has just 0 and 1
df["Region"] = df["Region"].astype(int)
```

```python
df.head()
```

```python
# any null vlaue
df[df.isnull().any(axis=1)]
```

```python
# drop the nan value
df = df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.loc[122]
# this is the header we have to remove this as well
```

```python
df = df.drop(122).reset_index(drop=True)
```

```python
df.iloc[122]
```
